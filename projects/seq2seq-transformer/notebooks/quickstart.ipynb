# Tiny offline execution smoke

This intentionally small synthetic training run checks the workflow, not accuracy. It uses unique disjoint date splits, saved-model inference, attention labels and reloaded evaluation. No dataset or pretrained tokenizer is downloaded.

```python
import os
from pathlib import Path
import uuid
project = Path.cwd()
if not (project/'configs').is_dir(): project = project.parent
assert (project/'configs').is_dir(), 'Launch from the project or notebooks folder'
os.chdir(project)
run = project/'artifacts'/('notebook-'+uuid.uuid4().hex[:12])
```

```python
from date_transformer.workflow import train,predict,evaluate_checkpoint
result = train(project/'configs/quickstart.json',run)
print('Tiny smoke metrics:',result['test'])
```

```python
prediction = predict(run/'best.pt','August 3, 2026',run/'example',attention=True)
print('Neural output:',prediction['prediction'])
print('Emitted EOS:',prediction['terminated'])
assert len(prediction['attention']) == len(prediction['attention_rows'])
metrics = evaluate_checkpoint(run/'best.pt',run/'reloaded-evaluation')
assert metrics == result['test']
```
