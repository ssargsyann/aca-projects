# ICU pipeline: runnable synthetic quickstart

This notebook uses **invented data and arbitrary labels**, not patient records. Its metrics do not measure ICU prediction quality. Install this project's `notebook` extras and select that Python environment before running all cells. No dataset download is required.


```python
from pathlib import Path
from uuid import uuid4
import pandas as pd
from icu_mortality.demo import run_demo

# Supports opening Jupyter from the project directory or its notebooks folder.
project = Path.cwd()
if project.name == 'notebooks':
    project = project.parent
if not (project / 'pyproject.toml').exists():
    raise RuntimeError('Start Jupyter from projects/icu-mortality-prediction or its notebooks directory.')
output = project / 'artifacts' / ('notebook-demo-' + uuid4().hex[:8])
summary = run_demo(output)
print(summary['notice'])
print('Outputs:', output)
```

## Inspect the invented input

Identifiers are excluded from modeling. Missing numeric values are imputed from the training partition. The target labels are deliberately arbitrary.


```python
pd.read_csv(output / 'synthetic_train.csv').head()
```

## Inspect demo validation and predictions

The metrics below are synthetic only. Example prediction rows are reused from the generated input to show how saved-model inference works; they are not another evaluation set.


```python
print('SYNTHETIC DEMO ONLY')
print('Validation ROC-AUC:', summary['metrics']['roc_auc'])
print('Validation average precision:', summary['metrics']['average_precision'])
pd.read_csv(output / 'predictions.csv').head()
```
