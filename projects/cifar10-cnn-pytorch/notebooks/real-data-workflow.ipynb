# Real CIFAR-10 workflow: short smoke run

Uses 1,000 training images, 200 validation images, and one CPU epoch. This is execution verification, not reproduction of the historical 81.56% result. Retrieve the data first, or point CIFAR10_DATA_ROOT at an existing cache.


```python
from pathlib import Path
from uuid import uuid4
import os, json
project = Path.cwd()
if project.name == 'notebooks':
    project = project.parent
if not (project / 'pyproject.toml').exists():
    raise RuntimeError('Start Jupyter from the CIFAR project or its notebooks folder.')
from cifar10_cnn.train import train
data_root = Path(os.environ.get('CIFAR10_DATA_ROOT', str(project / 'data')))
run = project / 'artifacts' / ('notebook-smoke-' + uuid4().hex[:8])
metadata = train(data_root, project / 'configs/smoke.json', run)
print(json.dumps(metadata, indent=2))
```

## Separate limited test evaluation

The trained smoke checkpoint is evaluated on a balanced 500-image official-test subset. This is not a full benchmark and is never used during training/checkpoint selection.


```python
from cifar10_cnn.evaluate import evaluate
evaluation_path = run / 'test-subset.json'
evaluated = evaluate(run / 'best.pt', data_root, evaluation_path, limit=500)
print('Test scope:', evaluated['split'], 'Samples:', evaluated['samples'])
print('Smoke accuracy:', evaluated['accuracy'])
```

## Training diagnostics

The plot is labeled with the run data kind. A one-epoch smoke plot cannot establish convergence.


```python
from cifar10_cnn.report import report
report(run, run / 'report', evaluation_path if 'evaluation_path' in globals() else None)
from IPython.display import Image, display
display(Image(filename=str(run / 'report/training-history.png')))
```
