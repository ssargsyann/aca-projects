```python
from pathlib import Path
from uuid import uuid4
import os, json
project = Path.cwd()
if project.name == 'notebooks':
    project = project.parent
if not (project / 'pyproject.toml').exists():
    raise RuntimeError('Start Jupyter from the CV Zoo project or its notebooks folder.')
from cv_zoo.geometry_workflow import demo
output = project / 'artifacts' / ('notebook-demo-' + uuid4().hex[:8])
result = demo(output)
for obj in result['objects']:
    print(obj['name'], 'cell:', obj['cell_col_row'], 'shape anchor:', obj['shape_winner'])
```
