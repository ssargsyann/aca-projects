# Four pretrained tasks on the real sample

Retrieve the image and pinned weights with the documented commands first, or configure CV_ZOO_IMAGE / CV_ZOO_WEIGHTS_DIR to existing caches. This notebook does not silently download files. Scores and masks are qualitative model predictions, not measured benchmark accuracy.


```python
from pathlib import Path
from uuid import uuid4
import os, json
project = Path.cwd()
if project.name == 'notebooks':
    project = project.parent
if not (project / 'pyproject.toml').exists():
    raise RuntimeError('Start Jupyter from the CV Zoo project or its notebooks folder.')
from cv_zoo.inference import inference, montage
image_path = Path(os.environ.get('CV_ZOO_IMAGE', str(project / 'assets/dog.jpg')))
weights_dir = Path(os.environ.get('CV_ZOO_WEIGHTS_DIR', str(project / 'artifacts/weights')))
output = project / 'artifacts' / ('notebook-real-' + uuid4().hex[:8])
result = inference(image_path, project / 'configs/inference.json', weights_dir, output / 'inference')
print('Input dimensions:', result['image_size'])
print('Top classification label:', result['tasks']['classification']['top_k'][0])
print('Detection outputs:', result['tasks']['detection']['count'])
print('Instance outputs:', result['tasks']['instance']['count'])
print('Semantic dimensions:', result['tasks']['semantic']['class_map_size'])
```

```python
from IPython.display import Image, display
montage(output / 'inference', output / 'montage.png')
display(Image(filename=str(output / 'montage.png')))
```

## Separate educational geometry

These are approximate teaching annotations from the original notebook. Shape-only anchor IoU and cell-centered IoU are distinct; neither is a trained detector accuracy metric. The supplied annotations require the original sample dimensions and content.


```python
from cv_zoo.geometry_workflow import geometry
geometry_result = geometry(image_path, project / 'configs/sample_geometry.json', output / 'geometry')
for obj in geometry_result['objects']:
    print(obj['name'], obj['cell_col_row'], obj['shape_winner'], obj['shape_ious'])
display(Image(filename=str(output / 'geometry/grid-anchors.png')))
```
